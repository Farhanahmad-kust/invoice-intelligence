```python
vendor_df = pd.read_sql_query("SELECT * FROM vendor_invoice", conn)
vendor_df.head()
```

```python
vendor_df[['Quantity','Freight','Dollars']].corr()
```

```python
vendor_df['freight_per_unit'] = vendor_df['Freight'] / vendor_df['Quantity']
```

```python
low_quantity = vendor_df['Quantity'].quantile(0.25)
high_quantity = vendor_df['Quantity'].quantile(0.75)

print(f"Low Quantity Threshold: {low_quantity}")
print(f"High Quantity Threshold: {high_quantity}")
```

```python
vendor_df.loc[vendor_df['Quantity'] < low_quantity,'freight_per_unit'].mean()
```

```python
vendor_df.loc[vendor_df['Quantity'] > high_quantity,'freight_per_unit'].mean()
```

```python
X = vendor_df[['Quantity','Dollars']]
Y = vendor_df['Freight']
```

```python
vendor_df.describe().round()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score,mean_absolute_error
```

```python
model1 = LinearRegression()
model1.fit(X_train,Y_train)

model2= DecisionTreeRegressor(max_depth=4, random_state=42)
model2.fit(X_train,Y_train)

model3= RandomForestRegressor(max_depth=5, random_state=42)
model3.fit(X_train,Y_train)

```

```python
def evaluate_model(model, X_test, Y_test, model_name):
    predictions = model.predict(X_test)
    mae = mean_absolute_error(Y_test, predictions)
    rmse = mean_squared_error(Y_test, predictions,)**0.5
    r2 = r2_score(Y_test, predictions)*100

    print(f"Model: {model_name} performance metrics:")
    print(f"Mean Absolute Error: {mae}")
    print(f"Root Mean Squared Error: {rmse}")
    print(f"R2 Score: {r2}")
```

```python
evaluate_model(model1, X_test, Y_test,"Linear Regression")
```

```python
evaluate_model(model2, X_test, Y_test,"Decision Tree Regressor")
```

```python
evaluate_model(model3, X_test, Y_test,"Random Forest Regressor")
```

```python
predictions = model1.predict(X_test)

plt.figure(figsize=(7, 5))
plt.scatter(Y_test, predictions, alpha=0.5)

minimum = min(Y_test.min(), predictions.min())
maximum = max(Y_test.max(), predictions.max())

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    color="red",
    linestyle="--"
)

plt.xlabel("Actual Freight")
plt.ylabel("Predicted Freight")
plt.title("Linear Regression: Actual vs Predicted Freight")
plt.show()
```
